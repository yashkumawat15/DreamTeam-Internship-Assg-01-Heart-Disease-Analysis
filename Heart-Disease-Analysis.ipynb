```python
def classify_age(age):
    if 18 <= age < 41:
        return "Young"
    elif 41 <= age <53:
        return "Middle"
    elif 53 <= age < 65:
        return "Senior"
    else:
        return "Super_Senior"

df['age_period'] = df['age'].apply(classify_age)
df.head()
```

```python
sns.countplot(data = df, x = 'age_period', hue = 'sex')
```
